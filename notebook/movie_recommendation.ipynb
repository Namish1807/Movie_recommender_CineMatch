```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(stop_words='english')
```

```python
tf_idf_matrix_overview = tfidf.fit_transform(df['overview'])
tf_idf_matrix_overview.shape
```

```python
tf_idf_matrix_genres = tfidf.fit_transform(df['genres'])
tf_idf_matrix_genres.shape
```

```python
tf_idf_matrix_keywords = tfidf.fit_transform(df['keywords'])
tf_idf_matrix_keywords.shape
```

```python
tf_idf_matrix_crew = tfidf.fit_transform(df['crew'])
tf_idf_matrix_crew.shape
```

```python
tf_idf_matrix_cast = tfidf.fit_transform(df['cast'])
tf_idf_matrix_cast.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_matrix_overview = cosine_similarity(tf_idf_matrix_overview)
similarity_matrix_genres = cosine_similarity(tf_idf_matrix_genres)
similarity_matrix_keywords = cosine_similarity(tf_idf_matrix_keywords)
similarity_matrix_crew = cosine_similarity(tf_idf_matrix_crew)
similarity_matrix_cast = cosine_similarity(tf_idf_matrix_cast)
```

```python
print(similarity_matrix_overview.shape)
print(similarity_matrix_genres.shape)
print(similarity_matrix_keywords.shape)
print(similarity_matrix_crew.shape)
print(similarity_matrix_cast.shape)
```

```python
final_similarity = (
    0.35 * similarity_matrix_overview +
    0.25 * similarity_matrix_genres +
    0.20 * similarity_matrix_keywords +
    0.15 * similarity_matrix_crew +
    0.05 * similarity_matrix_cast
)
```

```python
final_similarity.shape
```

```python
df.columns
```

```python
df = pd.merge(
    df,
    movies[['id','vote_average']],
    on = 'id',
    how='left'
)
```

```python
df.shape
```

```python
df.columns
```

```python
def recommend(movie_name, no_of_movies):
    movie_index = df[df["title"] == movie_name].index[0]

    similarity_scores = final_similarity[movie_index]

    recommended_movies = df[['title','genres','overview','vote_average']].iloc[
        similarity_scores.argsort()[::-1][1:no_of_movies+1]
    ]
    return recommended_movies
```

```python
recommend("Avatar", 10)
```
